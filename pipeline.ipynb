```python
# Pfade prüfen

import os

print("Aktueller Ordner:", os.getcwd())
print("Dateien:", os.listdir("."))

DB_FILE = "cine.db"
SCHEMA_FILE = "schema.sql"
```

```python
# DB vorbereiten

import os
import sqlite3

RESET_DB = True  # True = DB neu erstellen, False = vorhandene DB weiter benutzen

if RESET_DB and os.path.exists(DB_FILE):
    os.remove(DB_FILE)
    print("Alte DB gelöscht:", DB_FILE)

con = sqlite3.connect(DB_FILE)

with open(SCHEMA_FILE, "r", encoding="utf-8") as f:
    con.executescript(f.read())

con.commit()
con.close()
print("DB bereit:", DB_FILE)
```

```python
# Ergebnis anzeigen

import sqlite3

con = sqlite3.connect(DB_FILE)
cur = con.cursor()

cur.execute("SELECT COUNT(*) FROM film")
print("Anzahl Filme:", cur.fetchone()[0])

cur.execute("SELECT title, year FROM film ORDER BY year DESC LIMIT 5")
print("Beispiele:", cur.fetchall())

con.close()
```
